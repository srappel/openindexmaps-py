```python
import sqlite3
from pathlib import Path
```

```python
connection = sqlite3.connect(Path('../db/datastore.db'))
cursor = connection.cursor()
```

```python
# Create openindexmaps table
cursor.execute(
    """
CREATE TABLE IF NOT EXISTS openindexmaps (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    json_data TEXT,
    metadata TEXT,
    defaults TEXT,
    notes TEXT
)
"""
)
```

```python
def insert_openindexmap(name, json_data, metadata, defaults):
    cursor.execute(
        """
    INSERT INTO openindexmaps (name, json_data, metadata, defaults)
    VALUES (?, ?, ?, ?)
    """,
        (name, json_data, metadata, defaults),
    )

    connection.commit()
    connection.close()
```

```python
insert_openindexmap(
        "Test OIM",
        '{"type": "FeatureCollection"}',
        '{"id": "500"}',
        '{"publsiher": "Default Publishing House"}',
    )
```
